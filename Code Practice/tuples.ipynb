### Level 0 Absolute Beginner

```python
## Create a tuples...

tuple = ('Python', 'Java', 'C++')
print((tuple))
```

```python
## Tuple Element Access

languages = ('Python', 'Java', 'C++', 'Go')

# Print first  element:
print(languages[0])

# Prnt last element:
print(languages[-1])

# Print C++
print(languages [2])

# Print Java
print(languages[1])
```

### Level 1 ___ Beginner

```python
# Length and Indexing
```

```python
numbers = (10, 20, 30, 40, 50)

# get elements using (len)
print(len(numbers))

# Negative Indexing
print(numbers[-2:])
```

```python
# Tuple Slicing

fruits = ('Apple', 'Banana', 'Mango', 'Orange', 'Grapes', 'Guava')

# Using Slicing
# 1. Fitst 3 Fruit
print(fruits [0: 3])

# 2. Last 3 Fruit
print(fruits [-3:])

# 3. indexing 1 to 4
print(fruits [1:4])

# 4. Every Second Elements
print(fruits[1::2])
```

### Level 3 ___ Intermediate

```python
# Is tuple can Change?

numbers = (10, 20, 30)
numbers[1] = 50

print(numbers)


# We cannot change or add in Tuples. It's Immutable
```

```python
# Tuple Methods

numbers = (10, 20, 10, 30, 10, 40)

print(numbers.count(10))
print(numbers.index(30))

```

### Level 3 __Upper Intermediate

```python
# Tuple Unpacking

student = ('Polash', 25, 'Python')

# 3 different tuple unpacking:
name = student[0]
print(name)

age = student[1]
print(age)

language = student[2]
print(language)


## Challenge:
# Execute this code in one line
student = ('Polash', 25, 'Python')

name, age, language = student
print(name, age, language)
```

```python
## Extend Unpacking

numbers = (10, 20, 30, 40, 50, 60)

# Tuples unpacking Like a Variables
first, *middle, last = numbers


print(first)
print(middle)
print(last)
```

### Level 5 Advanced Challenge

```python
# Real-world Tuple Problem

# E-commerce Product___

product1 = ('Laptop', 85000, 5, 5)
product_name, price,quantity, discount_percentage = product1


# Discount_Amount.

original_price = price
discount_amount = price * quantity * discount_percentage / 100
discount_percentage = price * 5 /100
final_price = price - discount_percentage
total_price = price * quantity
final_amount = total_price - discount_amount

print('Product_name: ', product_name)
print('original Price:', original_price)
print('Discount Amount :', discount_amount)
print('Final Price Per Product: ', final_price )
print('Total Price:', total_price)
print('Final Payable Amount: ', final_amount)
```
