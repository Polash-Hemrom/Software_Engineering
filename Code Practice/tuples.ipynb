### Level 1 ___ Beginner

```python
# Length and Indexing
```

```python
numbers = (10, 20, 30, 40, 50)

# get elements using (len)
print(len(numbers))

# Negative Indexing
print(numbers[-2:])
```

```python
# Tuple Slicing

fruits = ('Apple', 'Banana', 'Mango', 'Orange', 'Grapes', 'Guava')

# Using Slicing
# 1. Fitst 3 Fruit
print(fruits [0: 3])

# 2. Last 3 Fruit
print(fruits [-3:])

# 3. indexing 1 to 4
print(fruits [1:4])

# 4. Every Second Elements
print(fruits[1::2])
```
