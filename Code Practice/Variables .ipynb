### Level 3 ___ Upper Intermediate

```python
## Variable Calculation

# Create appropriate variables and calculate:

product_cost = 1200
discount = 15

# Discount Amount Calculation:

discount_amount = product_cost * discount /100
print(discount_amount)


# Final Price
print(product_cost - discount_amount)
```
