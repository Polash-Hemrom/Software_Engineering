### Level 1 ___ Beginner

```python
# Add or Remove

languages = {'Python', 'Java', 'C++'}

# Add 'Go' in the languages
languages.add('Go')
print(languages)

# Remove 'Java' in the languages
languages.remove('Java')
print(languages)

# Check 'Java' in the languages
'Rust' in languages
```

```python
# 4. Set Membership

skills = {'Python', 'FastAPI', 'Docker', 'Git', 'Linux'}

# Using in to check the Word__ ::__

print('Python' in skills)
print('Javascript' in skills)
print('Docker' in skills)
```

### Level 2 ___ Intermediate

```python
## Union

python_students = {'Polash', 'James', 'Rinku', 'Peter'}
java_students = {'James', 'William', 'Peter', 'Jangu'}

print(python_students | java_students)
```

```python
## Intersection

print(python_students & java_students)
```

### Level 3 ___ Upper Intermediate

```python
backend = {'Python','Django', 'FastAPI', 'Node.js', 'PostgreSQL'}
frontend = {'HTML', 'CSS', 'JavaScript', 'React', 'Node.js'}


# Backend has technology that is not available in frontend.
print(backend - frontend)

# Frontend has technology that is not available in backend.
print(frontend - backend)
```

```python
# 8. Symmetric Difference

a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

# Numbers that are in only a or only b, but not in both sets.

print(a ^ b)
```

### Level 5 __ Advanced Real-World Challenge

```python
# 10. E-commerce Product Recommendation
# একটি e-commerce platform-এর দুইজন customer-এর purchased products:

customer_a = {
    'Laptop',
    'Mouse',
    'Keyboard',
    'Monitor',
    'Headphone'
}

customer_b = {
    'Laptop',
    'Mouse',
    'webcam',
    'Microphone',
    'Headphone'
}

# 1. Common products of both.
print(customer_a & customer_b)


# 2. Only products purchased by Customer A.
print(customer_a - customer_b)


# 3. Only products purchased by Customer B.
print(customer_b - customer_a)


# 4. How many types of unique products did the two of them buy in total?
print(customer_a | customer_b)

# 5. What products did Customer A buy that Customer B did not buy?

print(customer_a - customer_b)
```

### 🔥 Bonus Challenge:

```python
customer_a = {
    'Laptop',
    'Mouse',
    'Keyboard',
    'Monitor',
    'Headphone'
}


customer_b = {
    'Laptop',
    'Mouse',
    'Webcam',
    'Microphone',
    'Headphone'
}

new_customer = {
    'Laptop',
    'Mouse',
    'Keyboard'
}

all_product = customer_a | customer_b

recommendation = all_product - new_customer
print('Recommendation Products: ',recommendation)
```
