### Level 1 ___ Beginner

```python
# Add or Remove

languages = {'Python', 'Java', 'C++'}

# Add 'Go' in the languages
languages.add('Go')
print(languages)

# Remove 'Java' in the languages
languages.remove('Java')
print(languages)

# Check 'Java' in the languages
'Rust' in languages
```

```python
# 4. Set Membership

skills = {'Python', 'FastAPI', 'Docker', 'Git', 'Linux'}

# Using in to check the Word__ ::__

print('Python' in skills)
print('Javascript' in skills)
print('Docker' in skills)
```
