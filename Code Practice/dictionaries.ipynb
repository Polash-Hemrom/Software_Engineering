### Level 0 __ Absolute Beginner

```python
# 1. Create a dictionary of a student's information.

student = {
    'name' : 'Polash',
    'age' : 25,
    'country' : 'Bangladesh'
}

print(student)
```

```python
# Value Access
# Print the following three values ​​separately.

student = {
    'name' : 'Polash',
    'age': 25,
    'language' : 'Python'
}

print(student['name'])
print(student['age'])
print(student['language'])
```

### Level 1 __ Beginner

```python
# 3. Add, Update & Delete

student = {
    'name' : 'Polash',
    'age' : 25
}

# Add country name 'Bangladesh'

student['country'] = 'Bangladesh'
print(student)

# Update age 25 → 26.
student.update({'age': 26})
print(student)

# Delete "country".

student.pop('country')
print(student)
```
