### 🟠 Level 2 — Intermediate

```python
# 5. Dictionary Methods

product = {
    'name' : 'Laptop',
    'price' : 850000,
    'stock' : 10,
    'brand' : 'Apple'
}

# Print all 'keys'
for key in product.keys():
    print(key)

# Print all 'Values'
for value in product.values():
    print(value)

# সব key-value pairs

for key, value in product.items():
    print(key, value)

# How many items in Dictionaries
print(len(product))
```

```python
# 6. Dictionary to Calculation

product = {
    'name' : 'Laptop',
    'price' : 85000,
    'quantity': 3,
    'discount' : 10
}

# Calculate using values ​​from the dictionary.

# Original Total Pricet
total_price = product['price'] * product['quantity']
print(total_price)

# Discount Amount
discount_amount = total_price * product['discount'] /100
print(discount_amount)

# Final Payable

final_payable = total_price - discount_amount
print(final_payable)
```

### 🔵 Level 3 — Upper Intermediate

```python
# 7. Dictionary Iteration

scores = {
    'Polash' : 85,
    'Peter' : 100,
    'William' : 91,
    'Rinku' : 68,
    'Hemrom' : 88
}

# Print each student's name and score.
for key, value in scores.items():
    print(key, value)


# Find out who got the highest score.
print(max(scores, key = scores.get))

# Find out who got the lowest score.
print(min(scores, key = scores.get))

```

### Level 6 ___ Real-World Advanced Challenge

```python
# 10. Ecommerce Order System

order = {
    'customer' : 'Polash',
    'items':{
        'Laptop':{
            'price' : 850000,
            'quantity' : 1
        },

        'Mouse' : {
            'price' : 1500,
            'quantity' : 2

        }
    },
    'discount' : 10
}

# Step 1/ Find out the customer's name.
print('Customer name:', (order['customer']))


# Step 2/ Find out for each product.

items = order['items']

total_price = 0

for product_name, product_info in items.items():
    price = product_info['price']
    quantity = product_info['quantity']

    subtotal = price * quantity


    print('\nProduct :', product_name)
    print('Price :', price)
    print('Quantity :', quantity)
    print('Subtotal :', subtotal)


    total_price += subtotal


# Step 3: Total Price
print('\nTotal Price :', total_price)

# Step 4: Discount
discount_percentage = order['discount']

discount_amount = total_price * discount_percentage / 100
print('Discount :', discount_amount)


# Step 5: Final Payable Amount
final_amount = total_price - discount_amount

print('Final Payable Amount :', final_amount)
```
