### Level 0 __ Absolute Beginner

```python
# 1. Create a dictionary of a student's information.

student = {
    'name' : 'Polash',
    'age' : 25,
    'country' : 'Bangladesh'
}

print(student)
```

```python
# Value Access
# Print the following three values ​​separately.

student = {
    'name' : 'Polash',
    'age': 25,
    'language' : 'Python'
}

print(student['name'])
print(student['age'])
print(student['language'])
```

### Level 1 __ Beginner

```python
# 3. Add, Update & Delete

student = {
    'name' : 'Polash',
    'age' : 25
}

# Add country name 'Bangladesh'

student['country'] = 'Bangladesh'
print(student)

# Update age 25 → 26.
student.update({'age': 26})
print(student)

# Delete "country".

student.pop('country')
print(student)
```

### 🟠 Level 2 — Intermediate

```python
# 5. Dictionary Methods

product = {
    'name' : 'Laptop',
    'price' : 850000,
    'stock' : 10,
    'brand' : 'Apple'
}

# Print all 'keys'
for key in product.keys():
    print(key)

# Print all 'Values'
for value in product.values():
    print(value)

# সব key-value pairs

for key, value in product.items():
    print(key, value)

# How many items in Dictionaries
print(len(product))
```

```python
# 6. Dictionary to Calculation

product = {
    'name' : 'Laptop',
    'price' : 85000,
    'quantity': 3,
    'discount' : 10
}

# Calculate using values ​​from the dictionary.

# Original Total Pricet
total_price = product['price'] * product['quantity']
print(total_price)

# Discount Amount
discount_amount = total_price * product['discount'] /100
print(discount_amount)

# Final Payable

final_payable = total_price - discount_amount
print(final_payable)
```
