```python
# 1. Print numbers from 1 to 10 using a while loop.

number = 10
count = 0


while number > count:
    count += 1
    print(count)
```

```python
# 2. Print all even numbers from 1 to 50.

number = 50
count = 0

while number > count:
    count += 2
    print(count)
```

```python
# 3. Take a number from the user and count down to 0.

count = int(input('Enter a number: '))

while count > 0:
    count -= 1
    print(count)
```

```python
# 4. Take a number n from the user and calculate:

n = int(input('Enter a number: '))

count = n
total = 0

while count:
    total += count
    count -= 1


print(f'The sum of the first {n} natural number is {total}')
```
