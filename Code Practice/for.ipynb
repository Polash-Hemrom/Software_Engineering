### 🟢 Beginner

```python
# 1. Print Numbers
# 1 to 20


for number in range(1, 21):
    print(number)
```

```python
# 2. Print Even Numbers

for i in range(2, 52, 2):
    print(i)
```

```python
# 3. Sum of Numbers

num = 0

for i in range(1, 101):
    num += i

print(num)
```

```python
# 4. Multiplication Table

number = int(input('Enter your number: '))

for num in range(1, 11):
    print(number, 'X', num, '=', number * num)
```

#### 🟡 Intermediate

```python
## 5. Count Vowels


text = 'programming is fun'
vowels = 'aeiou'
vowels_count = 0

for char in text.lower():
    if char in vowels:
        vowels_count += 1

print(f'Number of vowels: {vowels_count}')
```

```python
## 6. Find the Largest Number

numbers = [12, 45, 7, 89, 23, 56]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest =  num

print(f'Find largest numbers: {largest}')
```

```python
## 7. Reverse String

text = 'Python'
reverse_txt = ''

for char in text:
    reverse_txt = char + reverse_txt

print(reverse_txt)
```

#### Challenge

```python
# 8. Find Duplicate Numbers

numbers = [2, 5, 3, 2, 8, 5, 9, 3, 1]

duplicates = []

for number in numbers:
    if numbers.count(number) > 1 and number not in duplicates:
        duplicates.append(number)


print(duplicates)
```
