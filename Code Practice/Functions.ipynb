#### Function Factory

```python
def introduce(name, age):
    return (f'My name is {name} and I am {age} years old.')


print(introduce(name = 'Polash', age = 30))
print(introduce(name = 'Peter', age = 25))
```

#### 🏦 Mini Banking System Step ny Step

```python
def withdraw(balance, withdraw_amount):
    if withdraw_amount > balance:
        return('Insufficient Balance')
    else:
        return('You can withdraw!')


print(withdraw(1500, 2000))
```

```python
def withdraw(balance, withdraw_amount):
    if withdraw_amount > balance:
        return 'Insufficient Balance'
    else:
        return balance - withdraw_amount

print(withdraw(1500, 500))
print(withdraw(1500, 2000))
```
