#### Level 2 ___ Parameters & Arguments

```python
def greet(name):
    print('Hello', name)


greet('Polash')
greet('William')
greet('Peter')
```

#### Level 3 ___ Return

```python
def add(a, b):
    return a + b


result = add(10, 20)
print(result)
print(result * 2)
print(result + 100)
```

#### Level 4 ___ Default Arguments

```python
def greet(name):
    print('Hello',name)


### We can default 

def greet(name = 'guest'):
    print('Hello', name)


greet()
```

#### Level 5 ___ Keyword Arguments

```python
def introduce(name, age):
    print(name, age)


introduce('Polash', 25)

introduce(age = 30, name = 'Peter')


# This is Keyword Arguments
```

#### Level 6 ___ *args

```python
def add(*numbers):
    print(numbers)


add(10, 20, 30, 40) # There Numbers is Tuple.
```

```python
def add(*numbers):
    total = 0

    for number in numbers:
        total += number

    return total

print(add(10, 20))
print(add(10, 20, 30))
print(add(1, 2, 3, 4, 5))
```

#### Level 7 ___ **kwargs

```python
# Now unlimited keyword arguments


def profile(**info):
    print(info)


profile(
    name = 'Polash',
    age = 25,
    skill = 'Python'
)
```

#### Level 8 ___ Scope

```python
# Now Dangerous Territory


x = 100         # This is Global Scope

def test():
    x  = 50     # This is Local Scope
    print(x)

test()
print(x)
```

#### Level 9 ___ Nested Functions

```python
def outer():

    def inner():
        print('Hello')

    inner()

outer()
```

#### Level 10 ___ Lambda

```python
# one line functions

square = lambda x: x * x

print(square(5))



# This is Normal Function:

def square(x):
    return (x * x)

print(square(6))
```
