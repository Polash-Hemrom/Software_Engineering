### The break Statement: Exiting a Loop Early

```python
number = 6

while number > 0:
    number -= 1

    if number == 2:
        break
    print(number)


print('Loop Ended!')
```

### The continue Statement: Skipping Tasks in an Iteration

```python
numbers = 6

while numbers > 0:
    numbers -= 1

    if numbers == 2:
        continue
    print(numbers)

print('Loop Ended')
    
```

```python
import random
import time


MAX_RETRIES = 5
attempts = 0


while attempts < MAX_RETRIES:
    attempts += 1

    print(f'Attempt {attempts}: Connecting to the server...')

    # Simulaating a connection scenario
    time.sleep(0.4)
    if random.choice([False, False, False, False, True]):
        print('Connection Successful!')
        break
    print('Connection failed. Retrying...')

else:
    print('All attempts failed. Unable to connect.')
```
